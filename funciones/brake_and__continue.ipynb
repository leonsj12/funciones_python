```python
# Variable para almacenar la suma de los números positivos
suma = 0
```

```python
# Ciclo while infinito (se romperá cuando se ingrese un número negativo)
while True:
    numero = int(input("Ingresa un número (negativo para salir): "))

    if numero >= 0:
        suma += numero  # Suma solo los números positivos
    else:
        # Muestra la suma total y el mensaje de despedida
        print(f"La suma de los números positivos es: {suma}")
        print("¡Gracias por usar el programa! 😊")
        break  # Rompe el ciclo si el número es negativo
```
